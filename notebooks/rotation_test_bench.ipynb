```python

bounding_box = [0,100,0,100,0,-100]
xmin, xmax, ymin, ymax, zmax, zmin = bounding_box
### HORIZONS ###

box = pv.Box(bounding_box)#.rotate_vector((0, 0, 1), 30, inplace=False)
rotated = pv.Box(bounding_box).rotate_vector((0, 0, 1), 30, inplace=False, point=(50,50,-50))
strat_over = pv.Plane((50,60,-50), i_size=851, j_size=851, direction=(0,0,1))
fault_over = pv.Plane((30,50,-50), i_size=801, j_size=801, direction=(1,0,0))
strat = whatamesh.clip_polydata(strat_over, xmin, xmax, ymin, ymax, zmin, zmax).rotate_vector((0, 0, 1), 30, inplace=False, point=(50,50,-50))
fault = whatamesh.clip_polydata(fault_over, xmin, xmax, ymin, ymax, zmin, zmax).rotate_vector((0, 0, 1), 30, inplace=False, point=(50,50,-50))

all_horizons = [strat]
volume_names = ["bottom", "top"]
vol_surface_names = ["Divider"]
all_faults = [fault]
fault_surface_names = ['fault']
model_name = "divided_vol"

plot = pv.Plotter()
plot.add_mesh(rotated,opacity=0.9)               
plot.add_mesh(strat,opacity=.5,color='blue')
plot.add_mesh(fault,opacity=0.6, color="red")

plot.show()
```
